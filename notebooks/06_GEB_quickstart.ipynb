# GEB quick start

Fit GEB to observations with two independent Beta(2, 2) coordinates on $[0,1]^2$. Evaluate the PPB posterior predictive density on a separate test sample.


```python
from pathlib import Path
import sys

root = Path.cwd().resolve()
if root.name == "notebooks":
    root = root.parent
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

import numpy as np
from methods.GEB import GEBTree
```

```python
rng = np.random.default_rng(2026)
x_train = rng.beta(2, 2, size=(5_000, 2))
x_test = rng.beta(2, 2, size=(1_000, 2))
```

```python
model = GEBTree(
    dimension_x=2,
    lower_global=np.zeros(2),
    upper_global=np.ones(2),
    alpha=1,
    dataset=x_train,
    m=32,
)
density = model.tree_joint_density_batch(x_test)
print(f"Mean test log score: {np.log(density).mean():.6f}")
```
